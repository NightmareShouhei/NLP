```python
path = kagglehub.dataset_download("seriousran/appletwittersentimenttexts")
dataframe = pd.read_csv(f"{path}/apple-twitter-sentiment-texts.csv")
dataframe.head()
```

```python
hashtags = re.findall(r"#\w+", " ".join(dataframe["text"]).lower())
print("Number of hashtags:", len(hashtags))
print("Top 10 hashtags:")
Counter(hashtags).most_common(10)
```

```python
text = "This year is 2021"
pattern = re.compile(r"\d+")
print(type(pattern))
text = pattern.sub("2022", text)
print(text)
# re.compile compiles a regular expression string into a reusable re.Pattern object
```

```python
text = "a 11 b 2 3 c 4"
print(re.split(r"\d+", text))
# re.split splits a string wherever the regular expression matches and returns the pieces as a list
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("I'm enjoying the NLP course!")
for token in doc:
    print(token.text)

# spacy.load loads a trained pipeline and returns an nlp object
```

```python
print("spaCy:", [token.text for token in doc])
print("NLTK: ", nltk.word_tokenize("I'm enjoying the NLP course!"))
```
