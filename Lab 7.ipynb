```python
# Sentiment classification: Encoder-only (BERT, RoBERTa)
# Machine translation: Encoder-decoder (T5, BART)
# Open-ended text generation: Decoder-only (GPT)
# Named Entity Recognition: Encoder-only (BERT, RoBERTa)
# Text summarization: Encoder-decoder (T5, BART)
```

```python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
sentences = ["The movie was excellent.", "The lecture was difficult to understand.", "I enjoyed learning about transformers."]

for sentence in sentences:
    result = classifier(sentence)[0]
    print(sentence, "->", result["label"], round(result["score"], 4))
```

```python
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

paragraph = "Natural language processing (NLP) is a field of artificial intelligence that helps computers understand, interpret and generate human language. It combines linguistics with machine learning to analyze text and speech. NLP is used in many everyday applications such as search engines, spam filters, voice assistants, machine translation and chatbots. In recent years, large language models based on transformers have greatly improved the quality of NLP systems."
output = model.generate(**tokenizer("summarize: " + paragraph, return_tensors="pt"), max_new_tokens=50)
print(tokenizer.decode(output[0], skip_special_tokens=True))
```

```python
# A decoder-only model generates text by predicting the next token from all the tokens before it, which is how a chatbot writes a reply.
# The whole conversation is given to it as one sequence, so it uses everything said so far as context.
# Since it is pre-trained on huge amounts of text, it can give fluent, open-ended answers.
```
