```python
nlp = spacy.load("en_core_web_sm")
sen = nlp("I like watching football matches. I do not know how to play it though")
print(sen[12].pos_)
print(sen[12].tag_)
print(spacy.explain(sen[12].tag_))
```

```python
for word in sen:
    print(word.text, word.pos_, word.tag_, spacy.explain(word.tag_))
```

```python
sentences = [("Did you search on Google?", 4), ("Let us google it.", 2), ("The has placed the food on the table", 7), ("She plans to table a motion for debate in Parliament", 3)]

for text, i in sentences:
    word = nlp(text)[i]
    print(text, "->", word.pos_, word.tag_, spacy.explain(word.tag_))

# Google: a proper noun when it is the company name, but a verb when it means to search
# table: a noun when it is the piece of furniture, but a verb when it means to propose a motion
```
