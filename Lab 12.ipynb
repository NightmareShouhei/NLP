```python
# Hamza Alsihati
# 2230006810

import torch
import pandas as pd
from transformers import AutoTokenizer, AutoModel
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased", attn_implementation="eager")

layers = model.config.num_hidden_layers
heads = model.config.num_attention_heads
print("transformer layers:", layers)
print("attention heads per layer:", heads)
```

```python
print("Total attention heads:", layers * heads)
```

```python
inputs = tokenizer("Transformers use self-attention to understand language.", return_tensors="pt")
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

with torch.no_grad():
    outputs = model(**inputs, output_attentions=True)

attention = outputs.attentions[2][0, 1]
print("Attention shape:", attention.shape)
pd.DataFrame(attention.numpy(), index=tokens, columns=tokens).round(2)
```
